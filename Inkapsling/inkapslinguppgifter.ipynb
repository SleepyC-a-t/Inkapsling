### Övning 3 — Metoder som kontrollerar

Lägg till två metoder i `Spelare`:

- `ta_skada(skada)` minskar hälsan med `skada`.
- `hela(mangd)` ökar hälsan med `mangd`.

Metoderna ska skydda hälsan:

- Om `skada` eller `mangd` är negativt ska ett felmeddelande skrivas ut och hälsan inte ändras.
- Hälsan får **aldrig bli lägre än 0**. Blir den det ska den sättas till 0.
- Hälsan får **aldrig bli högre än 100**. Blir den det ska den sättas till 100.

Behåll `visa_status()` från övning 2. Testa med:

```python
spelare = Spelare("Alex")
spelare.ta_skada(30)
spelare.hela(50)
spelare.ta_skada(-10)
spelare.ta_skada(500)
spelare.visa_status()
```

Rätt svar på sista raden är `Alex har 0 i hälsa.`

*Tips: titta på exempel 3.*

````python
# Skriv din kod här
class Player:
    def __init__(self, name, health=100):
        self.name = name
        self._health = health

    def view_status(self):
        print(f"{self.name} has {self._health} health remaining")

    def take_damage(self, damage):
        if damage <= 0:
            print("Health has not changed")
        elif (self._health - damage) < 0:
            self._health = 0
        else:
            self._health -= damage
            print(f"{self.name} took {damage} damage and has {self._health} health remaining")

    def healing(self, heal):
        if heal < 0:
            print("Health has not changed")
        elif (self._health + heal) > 100:
            self._health = 100
        else:
            self._health += heal
            print(f"{self.name} has recieved {heal} healing and has {self._health} health now")

player1 = Player("player1")
player1.take_damage(30)
player1.healing(50)
player1.take_damage(-10)
player1.take_damage(500)
player1.view_status()
````

### Övning 6 — `@property`

Byt ut gettern `hamta_halsa()` mot en **property** som heter `halsa`, så att man kan skriva:

```python
print(spelare.halsa)
```

utan parenteser.

Testa sedan att skriva `spelare.halsa = 999` inuti ett `try`/`except AttributeError` och skriv ut ett meddelande när det nekas.

*Tips: titta på exempel 6. Propertyn heter* `halsa` *och attributet* `_halsa`*.*

````python
# Skriv din kod här
class Player:
    def __init__(self, name, health=100):
        self.name = name
        self._health = health

    def view_status(self):
        print(f"{self.name} has {self._health} health remaining")

    @property
    def health(self):
        return self._health

player1 = Player("player1")
print(f"Health is {player1.health}")

try:
    player1.health = 999
except:
    print("Invalid health amount")
print(f"Health is still {player1.health}")
````

### Övning 7 — `@property` med setter

Lägg till en **setter** till propertyn `halsa` från övning 6. Settern ska bara godkänna värden mellan **0 och 100** (båda inräknade). Andra värden ska ge ett felmeddelande och lämna hälsan oförändrad.

Testa:

```python
spelare.halsa = 50      # godkänns
spelare.halsa = 150     # nekas
spelare.halsa = -20     # nekas
print(spelare.halsa)    # ska skriva ut 50
```

*Tips: titta på exempel 7. Ett smidigt sätt att skriva villkoret är* `0 <= nytt_varde <= 100`*.*

````python
# Skriv din kod här
class Player:
    def __init__(self, name, health=100):
        self._name = name
        self._health = health

    def view_status(self):
        print(f"{self._name} has {self._health} health remaining")

    @property
    def health(self):
        return self._health

    @health.setter
    def health(self, new_health_amount):
        if 0 <= new_health_amount <= 100:
            self._health = new_health_amount

player1 = Player("player1")
player1.health = 50
player1.health = 150
player1.health = -20
player1.view_status()
````

### Övning 8 — Beräknad property

Lägg till en property `lever` som returnerar `True` om spelarens hälsa är större än 0, annars `False`.

Använd sedan `lever` i ett villkor:

```python
if spelare.lever:
    print("Spelaren lever!")
else:
    print("Game over!")
```

Testa både när spelaren har full hälsa och efter att du satt hälsan till 0 (med settern från övning 7).

*Tips: titta på exempel 8.*

````python
# Skriv din kod här
class Player:
    def __init__(self, name, health=100):
        self._name = name
        self._health = health

    def view_status(self):
        print(f"{self._name} has {self._health} health remaining")

    @property
    def health(self):
        return self._health

    @health.setter
    def health(self, new_health_amount):
        if 0 <= new_health_amount <= 100:
            self._health = new_health_amount

    @property
    def alive(self):
        if self._health > 0:
            return True
        else:
            return False

player1 = Player("player1")

if player1.alive:
    print("The player is alive!")
else:
    print("Game over!")
player1.health = 1
if player1.alive:
    print("The player is alive!")
else:
    print("Game over!")
player1.health = 0
if player1.alive:
    print("The player is alive!")
else:
    print("Game over!")
````

### Övning 10 — Inkapsling och arv: `Krigare`

Skapa en klass `Krigare` som **ärver** från `Spelare`. En krigare har rustning som skyddar mot skada.

`Spelare` ska ha:

- `namn`, `_halsa` (börjar på 100) och propertyn `halsa` (bara getter räcker)
- metoden `ta_skada(skada)` som minskar hälsan men aldrig under 0

`Krigare` ska:

- ta emot `namn` och `rustning` i konstruktorn och använda `super().__init__(namn)`
- spara rustningen som det skyddade attributet `_rustning`
- **skriva över** `ta_skada(skada)` så att skadan först minskas med rustningen. Skadan får inte bli negativ, så minsta skada är 0. Anropa sedan `super().ta_skada(...)` med den minskade skadan.

Testa:

```python
vanlig = Spelare("Alex")
krigare = Krigare("Conan", 15)
vanlig.ta_skada(40)
krigare.ta_skada(40)
krigare.ta_skada(10)
print(vanlig.namn, vanlig.halsa)      # Alex 60
print(krigare.namn, krigare.halsa)    # Conan 75
```

*Tips: titta på exempel 10, och på exempel 8 i arvsavsnittet om* `super()`*.*

````python
# Skriv din kod här
class Player:
    def __init__(self, name, health=100):
        self._name = name
        self._health = health

    @property
    def get_health(self):
        return self._health

    @property
    def get_name(self):
        return self._name

    def take_damage(self, damage):
        if damage <= 0:
            print("Health has not changed")
        elif (self._health - damage) < 0:
            self._health = 0
            print(f"{self._name} took {damage} damage and has {self._health} health remaining")
        else:
            self._health -= damage
            print(f"{self._name} took {damage} damage and has {self._health} health remaining")

class Warrior(Player):
    def __init__(self, name, armor, health=100):
        super().__init__(name, health)
        self._armor = armor

    def take_damage(self, damage):
        # Skadan kan i sig bli negativ men det gör inget för att det kontrolleras redan i standard take_damage checken
        damage -= self._armor
        super().take_damage(damage)

player1 = Player("player1")
warrior1 = Warrior("warrior1", 15)
player1.take_damage(40)
warrior1.take_damage(40)
warrior1.take_damage(10)
print(player1.get_name, player1.get_health)
print(warrior1.get_name, warrior1.get_health)
````
